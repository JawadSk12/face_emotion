```python
def load_face_images_for_lbph(base_dir, target_size=(200, 200)):
    face_samples = []   # list of 2D uint8 arrays
    labels = []         # list of ints
    label_map = {}      # id -> name
    current_id = 0

    if not os.path.isdir(base_dir):
        raise FileNotFoundError(f"Known faces directory not found: {base_dir}")

    print(f"[INFO] Scanning folder: {os.path.abspath(base_dir)}")

    for person_name in os.listdir(base_dir):
        person_folder = os.path.join(base_dir, person_name)
        if not os.path.isdir(person_folder):
            continue

        print(f"[INFO] Loading images for person: {person_name}")
        label_map[current_id] = person_name

        for img_name in os.listdir(person_folder):
            if not img_name.lower().endswith((".jpg", ".jpeg", ".png")):
                continue

            img_path = os.path.join(person_folder, img_name)
            img = cv2.imread(img_path)
            if img is None:
                print(f"[WARN] Unable to read image: {img_path}")
                continue

            gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            gray_resized = cv2.resize(gray, target_size)

            face_samples.append(gray_resized)
            labels.append(current_id)

        current_id += 1

    print(f"[INFO] Total images loaded: {len(face_samples)}")
    print(f"[INFO] Total unique persons: {len(label_map)}")

    return face_samples, labels, label_map
```

```python
known_dir = config["paths"]["known_faces_dir"]  # should be "data/known_faces"
faces, ids, label_map = load_face_images_for_lbph(known_dir)

print("faces count:", len(faces))
print("ids count:", len(ids))
print("label_map:", label_map)
```
